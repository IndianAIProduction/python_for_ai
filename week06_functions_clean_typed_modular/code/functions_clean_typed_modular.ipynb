## CLASS 37 — Recursion

📖 Technical definition — Recursion: Recursion is a technique in which a function solves a problem by calling itself on a smaller version of that problem. It requires a base case that stops the calls and a recursive case that reduces the problem toward the base case.

```python
def countdown(n):
    print(n)

    if n == 1:
        print("Countdown completed")
        return 

    countdown(n-1)
```

```python
countdown(5)
```

```python
countdown(100)
```

👨‍🏫 Concept 2 — factorial

5 * 4 * 3 * 2 *1 = 5!
4 * 3 * 2 *1 = 4!
10*9.......*1 = 10!

```python
def factorial(n):

    if n == 1:
        return 1

    return n * factorial(n-1) ##  2 * factorial(1) >> 2 * 1 = 2
```

```python
factorial(2)
```

```python
factorial(5)
```

👨‍🏫 Concept 3 — nested data

sum of each element >> [1, 2, [3, 4, [5, 6]], 7] >> 28

```python
def sum_nest_list(list_data): ##  7, [5, 6]                    ##[3, 4, [5, 6]]              #1, 2, [3, 4, [5, 6]], 7]
    total = 0
    
    for item in list_data: #                                1, 2, [3, 4, [5, 6]]
        if isinstance(item, list):
            total = total + sum_nest_list(item)
        else:
            total = total + item # 21
        
    return total


    
```

```python
sum_nest_list([1, 2, [3, 4, [5, 6]], 7])
```

📖 Technical definition — Type hint: A type hint is an optional annotation that declares the expected type of a parameter, variable, or return value. Python does not enforce it at runtime, but it documents intent and enables tools (and language models) to check and understand the code.

👨‍🏫 Concept 1 — lambda (mini one-line function)

📖 Technical definition — Lambda: A lambda is a small, anonymous function defined in a single expression. It takes arguments and returns the value of that expression automatically, without a name or an explicit return statement.

```python
def square(x: int) -> int:
    return x ** 2
```

👨‍🏫 Concept 2 — map() (har item par function lagao)

📖 Technical definition — map() and filter(): map() applies a given function to every item of an iterable and yields the transformed results. filter() keeps only the items for which a given function returns a truthy value. Both return lazy iterators, so list()) to see the values.

map(function, items)

```python
nums = [3,4,5,1,2]

sqr_list = list(map(lambda x: x ** 2, nums))
```

```python
sqr_list
```
