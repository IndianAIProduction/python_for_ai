👨‍🏫 Concept 2 — __init__ (object bante hi chalne wala setup)

📖 Technical definition — __init__ (constructor): __init__ is a special method that Python calls automatically right after a new object is created. It initialises the object's starting attribute values from the arguments passed when the object is constructed.

```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks
```

👨‍🏫 Concept 4 — attributes vs methods

📖 Technical definition — Attribute vs method: An attribute is a piece of data stored on an object (its state). A method is a function defined inside a class that operates on the object (its behaviour). Together they make up what an object "has" and what it "does."

- Attribute = object ka DATA (self.name, self.age). Noun.
- Method = object ka KAAM (bark(), increase()). Verb. (Method = function jo class ke andar hai.)

👨‍🏫 Concept 2 — _ (underscore) convention — "yeh private hai"

```python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount 
        return print("Available balance is", self._balance)

    def withdraw(self, amount):
        if amount > 0:
            self._balance -= amount
        return print("Available balance is", self._balance)

    def get_balanace(self):
        return self._balance

```

👨‍🏫 Concept 3 — validation (methods galat data rokte hain)

```python
class BankAccount:
    """
    A class to represent a bank account.
    Attributes:
        balance (int): The balance of the bank account. 
        Methods:
            deposit(amount: int) -> None: Deposit money into the account.
            withdraw(amount: int) -> None: Withdraw money from the account.
            get_balance() -> int: Get the current balance of the account.
    """
    def __init__(self, balance: int):
        self._balance = balance

    def deposit(self, amount: int) -> None:
        if amount <= 0:
            return print("Invalid amount")

        if amount > 0:
            self._balance += amount 
        return print("Available balance is", self._balance)

    def withdraw(self, amount: int) -> None:
        if amount <= 0:
            return print("Invalid amount")

        if amount > 0:
            self._balance -= amount
        return print("Available balance is", self._balance)

    def get_balanace(self) -> int:
        return self._balance
```

```python
bank_account_1 = BankAccount(3000)
```

```python
bank_account_1.deposit(-500)
```

```python
bank_account_1._balance = 500000
```

```python
bank_account_1._balance
```

👨‍🏫 Concept 4 — __ (double underscore) — aur strict

```python
class BankAccount:
    """
    A class to represent a bank account.
    Attributes:
        balance (int): The balance of the bank account. 
        Methods:
            deposit(amount: int) -> None: Deposit money into the account.
            withdraw(amount: int) -> None: Withdraw money from the account.
            get_balance() -> int: Get the current balance of the account.
    """
    def __init__(self, balance: int):
        self.__balance = balance

    def deposit(self, amount: int) -> None:
        if amount <= 0:
            return print("Invalid amount")

        if amount > 0:
            self.__balance += amount 
        return print("Available balance is", self.__balance)

    def withdraw(self, amount: int) -> None:
        if amount <= 0:
            return print("Invalid amount")

        if amount > self.__balance:
            return print("Insufficient balance")
            
        if amount > 0:
            self.__balance -= amount
        return print("Available balance is", self.__balance)

    def get_balanace(self) -> int:
        return self.__balance
```

```python
bank_account_1 = BankAccount(300)
```

```python
bank_account_1.__balance
```

```python
bank_account_1.get_balanace()
```

```python
bank_account_1.withdraw(500)
```

```python
bank_account_1.withdraw(300)
```
